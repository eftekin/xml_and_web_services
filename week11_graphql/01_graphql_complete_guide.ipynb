## Part 1: Introduction to GraphQL

### What is GraphQL?

**GraphQL** is a query language for APIs and a runtime for fulfilling those queries with your existing data. Created by Facebook in 2012 and open-sourced in 2015.

### Key Concepts

```python
print("=" * 80)
print("🔍 GRAPHQL vs REST: THE FUNDAMENTAL DIFFERENCE")
print("=" * 80)

print("\n📱 REST API Approach:")
print("   Problem: You need user data + their posts + comments")
print("""   
   GET /users/1              → Get user
   GET /users/1/posts        → Get user's posts  
   GET /posts/1/comments     → Get post comments
   GET /posts/2/comments     → Get more comments
   ...
   
   Result: Multiple HTTP requests (N+1 problem)
   Issue: Over-fetching (getting more data than needed)
""")

print("\n🚀 GraphQL Approach:")
print("   Solution: ONE request, ASK for EXACTLY what you need")
print("""   
   POST /graphql
   
   Query:
   {
     user(id: 1) {
       name
       email
       posts {
         title
         comments {
           text
           author { name }
         }
       }
     }
   }
   
   Result: ONE HTTP request, EXACTLY the data you specified
""")

print("=" * 80)
print("💡 KEY INSIGHT:")
print("   REST: Multiple endpoints, server decides response structure")
print("   GraphQL: One endpoint, CLIENT decides response structure")
print("=" * 80)
```

### Core GraphQL Operations

```python
print("=" * 80)
print("📋 THREE CORE GRAPHQL OPERATIONS")
print("=" * 80 + "\n")

operations = [
    (
        "1. Query",
        "Read/Fetch Data",
        "Like GET in REST",
        """query {
  user(id: 1) {
    name
    email
  }
}"""
    ),
    (
        "2. Mutation",
        "Write/Modify Data",
        "Like POST/PUT/DELETE in REST",
        """mutation {
  createUser(name: "John", email: "john@example.com") {
    id
    name
  }
}"""
    ),
    (
        "3. Subscription",
        "Real-time Updates",
        "WebSocket-based real-time data",
        """subscription {
  messageAdded {
    id
    text
    author
  }
}"""
    )
]

for operation, purpose, equivalent, example in operations:
    print(f"\n{operation}")
    print(f"Purpose: {purpose}")
    print(f"REST Equivalent: {equivalent}")
    print(f"\nExample:")
    print(example)
    print("-" * 80)
```

## Part 2: GraphQL Queries

### Basic Query Syntax

```python
print("=" * 80)
print("📝 GRAPHQL QUERY SYNTAX")
print("=" * 80 + "\n")

# Example 1: Simple query
print("1️⃣  SIMPLE QUERY - Get user name\n")
simple_query = """{
  user(id: 1) {
    name
  }
}"""
print("Query:")
print(simple_query)

response1 = """{
  "data": {
    "user": {
      "name": "John Doe"
    }
  }
}"""
print("\nResponse:")
print(response1)

# Example 2: Multiple fields
print("\n" + "="*80)
print("\n2️⃣  MULTIPLE FIELDS - Get specific user data\n")
multi_field_query = """{
  user(id: 1) {
    id
    name
    email
    age
  }
}"""
print("Query:")
print(multi_field_query)

# Example 3: Nested queries
print("\n" + "="*80)
print("\n3️⃣  NESTED QUERY - User with posts\n")
nested_query = """{
  user(id: 1) {
    name
    posts {
      title
      content
      createdAt
    }
  }
}"""
print("Query:")
print(nested_query)

# Example 4: Arguments
print("\n" + "="*80)
print("\n4️⃣  WITH ARGUMENTS - Filter and limit\n")
args_query = """{
  posts(limit: 10, orderBy: "createdAt", order: DESC) {
    id
    title
    author {
      name
    }
  }
}"""
print("Query:")
print(args_query)

# Example 5: Aliases
print("\n" + "="*80)
print("\n5️⃣  ALIASES - Query same field with different arguments\n")
alias_query = """{
  activeUsers: users(status: ACTIVE) {
    name
  }
  inactiveUsers: users(status: INACTIVE) {
    name
  }
}"""
print("Query:")
print(alias_query)

# Example 6: Fragments
print("\n" + "="*80)
print("\n6️⃣  FRAGMENTS - Reusable query pieces\n")
fragment_query = """fragment UserInfo on User {
  id
  name
  email
}

{
  user1: user(id: 1) {
    ...UserInfo
  }
  user2: user(id: 2) {
    ...UserInfo
  }
}"""
print("Query:")
print(fragment_query)

print("\n" + "="*80)
```

## Part 3: GraphQL Schema

### Schema Definition Language (SDL)

```python
print("=" * 80)
print("📐 GRAPHQL SCHEMA DEFINITION")
print("=" * 80 + "\n")

schema_example = '''
# Scalar Types (built-in)
# Int, Float, String, Boolean, ID

# Object Type
type User {
  id: ID!              # ! means non-nullable (required)
  name: String!
  email: String!
  age: Int
  posts: [Post!]!      # Array of Posts (array and items are required)
  createdAt: String
}

type Post {
  id: ID!
  title: String!
  content: String
  published: Boolean!
  author: User!        # Relationship to User
  comments: [Comment!]!
}

type Comment {
  id: ID!
  text: String!
  author: User!
  post: Post!
}

# Query Type (read operations)
type Query {
  user(id: ID!): User
  users(limit: Int, offset: Int): [User!]!
  post(id: ID!): Post
  posts(published: Boolean): [Post!]!
}

# Mutation Type (write operations)
type Mutation {
  createUser(name: String!, email: String!): User!
  updateUser(id: ID!, name: String, email: String): User
  deleteUser(id: ID!): Boolean!
  
  createPost(title: String!, content: String, authorId: ID!): Post!
  publishPost(id: ID!): Post
}

# Subscription Type (real-time updates)
type Subscription {
  postAdded: Post!
  commentAdded(postId: ID!): Comment!
}

# Input Type (for complex inputs)
input CreateUserInput {
  name: String!
  email: String!
  age: Int
}

# Enum Type
enum UserRole {
  ADMIN
  USER
  GUEST
}
'''

print("Complete Schema Example:")
print(schema_example)

print("\n" + "="*80)
print("🔑 SCHEMA KEY POINTS:")
print("""
  1. ! = Required (non-nullable)
  2. [Type] = Array 
  3. [Type!]! = Required array of required items
  4. type Query = Read operations
  5. type Mutation = Write operations
  6. type Subscription = Real-time updates
  7. Scalar types: Int, Float, String, Boolean, ID
  8. Custom types: User, Post, etc.
""")
print("="*80)
```

## Part 4: GraphQL Mutations

Mutations modify data on the server:

```python
print("=" * 80)
print("✏️  GRAPHQL MUTATIONS (Write Operations)")
print("=" * 80 + "\n")

# Create
print("1️⃣  CREATE - Add new user\n")
create_mutation = '''mutation {
  createUser(name: "Alice Smith", email: "alice@example.com") {
    id
    name
    email
    createdAt
  }
}'''
print("Mutation:")
print(create_mutation)

print("\nExpected Response:")
print('''{
  "data": {
    "createUser": {
      "id": "123",
      "name": "Alice Smith",
      "email": "alice@example.com",
      "createdAt": "2026-02-11T10:00:00Z"
    }
  }
}''')

# Update
print("\n" + "="*80)
print("\n2️⃣  UPDATE - Modify existing user\n")
update_mutation = '''mutation {
  updateUser(id: "123", name: "Alice Johnson") {
    id
    name
    email
  }
}'''
print("Mutation:")
print(update_mutation)

# Delete
print("\n" + "="*80)
print("\n3️⃣  DELETE - Remove user\n")
delete_mutation = '''mutation {
  deleteUser(id: "123")
}'''
print("Mutation:")
print(delete_mutation)

# Variables
print("\n" + "="*80)
print("\n4️⃣  USING VARIABLES (Best Practice)\n")
variable_mutation = '''mutation CreateUser($name: String!, $email: String!) {
  createUser(name: $name, email: $email) {
    id
    name
    email
  }
}'''

variables = '''{
  "name": "Bob Wilson",
  "email": "bob@example.com"
}'''

print("Mutation:")
print(variable_mutation)
print("\nVariables:")
print(variables)

print("\n" + "="*80)
print("💡 MUTATION BEST PRACTICES:")
print("""
  ✓ Use variables instead of inline values
  ✓ Return the modified data
  ✓ Include error fields in response type
  ✓ Make mutations explicit and descriptive
  ✓ Consider batch mutations for multiple operations
""")
print("="*80)
```

## Part 5: GraphQL in Python

### Using GraphQL Client

```python
# Example: Consuming a GraphQL API in Python

import requests
import json

def execute_graphql_query(endpoint, query, variables=None):
    """
    Execute a GraphQL query
    
    Args:
        endpoint: GraphQL endpoint URL
        query: GraphQL query string
        variables: Optional variables dictionary
    
    Returns:
        Response data or None if error
    """
    headers = {
        'Content-Type': 'application/json',
    }
    
    payload = {
        'query': query,
    }
    
    if variables:
        payload['variables'] = variables
    
    try:
        response = requests.post(
            endpoint,
            headers=headers,
            json=payload,
            timeout=10
        )
        response.raise_for_status()
        
        result = response.json()
        
        if 'errors' in result:
            print("GraphQL Errors:")
            for error in result['errors']:
                print(f"  - {error['message']}")
            return None
        
        return result['data']
        
    except requests.exceptions.RequestException as e:
        print(f"Request error: {e}")
        return None

# Example usage (with a public GraphQL API)
print("=" * 80)
print("🐍 GRAPHQL IN PYTHON")
print("=" * 80 + "\n")

# Example query
query = '''
query GetUser($userId: ID!) {
  user(id: $userId) {
    id
    name
    email
    posts {
      title
      published
    }
  }
}
'''

variables = {
    'userId': '1'
}

print("Query Structure:")
print(query)
print("\nVariables:")
print(json.dumps(variables, indent=2))

print("\n" + "="*80)
print("\n💡 PYTHON GRAPHQL LIBRARIES:")
print("""
  1. gql - GraphQL client library
     pip install gql
     
  2. graphene - Build GraphQL schemas
     pip install graphene
     
  3. strawberry - Modern GraphQL library  
     pip install strawberry-graphql
     
  4. ariadne - Schema-first GraphQL
     pip install ariadne
""")
print("="*80)
```

## Part 6: GraphQL vs REST

### Detailed Comparison

```python
from tabulate import tabulate

print("=" * 80)
print("⚖️  GRAPHQL vs REST COMPARISON")
print("=" * 80 + "\n")

comparison = [
    ["Endpoints", "Multiple (/users, /posts)", "Single (/graphql)"],
    ["Data Fetching", "Fixed response structure", "Client specifies fields"],
    ["Over-fetching", "Common problem", "Not an issue"],
    ["Under-fetching", "Requires multiple requests", "Get all in one request"],
    ["Versioning", "Often needs /v1, /v2", "No versioning needed"],
    ["Documentation", "Separate (Swagger, etc.)", "Self-documenting (introspection)"],
    ["Caching", "Easy (HTTP caching)", "More complex"],
    ["Learning Curve", "Lower", "Higher"],
    ["Type System", "Not built-in", "Strong typing built-in"],
    ["Real-time", "WebSockets needed", "Built-in (subscriptions)"],
]

print(tabulate(comparison, headers=["Aspect", "REST", "GraphQL"], tablefmt="grid"))

print("\n" + "="*80)
print("✅ USE GRAPHQL WHEN:")
print("""
  • Mobile apps (bandwidth matters)
  • Complex, nested data requirements
  • Rapid frontend development
  • Multiple clients with different needs
  • Real-time features needed
  • You want strong typing
""")

print("✅ USE REST WHEN:")
print("""
  • Simple, CRUD-focused APIs
  • HTTP caching is important
  • File uploads/downloads
  • Team is familiar with REST
  • Simple client applications
  • Infrastructure optimized for REST
""")

print("💡 THEY CAN COEXIST:")
print("""
  Many organizations use both:
  • REST for simple operations
  • GraphQL for complex querying
  • REST for file handling
  • GraphQL for data aggregation
""")
print("="*80)
```
