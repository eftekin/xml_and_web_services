## 1. Anatomy of an XML Document

Let's start with a complete, well-formed XML document and break down every part:

````python
# A complete XML document example

xml_document = '''<?xml version="1.0" encoding="UTF-8"?>
<!-- This is a comment: Book catalog example -->
<catalog>
    <book id="bk101" category="programming">
        <title>XML Developer's Guide</title>
        <author>Gambardella, Matthew</author>
        <price currency="USD">44.95</price>
        <publish_date>2000-10-01</publish_date>
        <description>An in-depth look at creating applications with XML.</description>
    </book>
</catalog>'''

print("=" * 70)
print("📄 COMPLETE XML DOCUMENT")
print("=" * 70)
print(xml_document)
print("\n" + "=" * 70)

# Let's identify each component
print("\n🔍 DOCUMENT COMPONENTS:\n")

components = [
    ("Line 1", "XML Declaration", "<?xml version=\"1.0\" encoding=\"UTF-8\"?>"),
    ("Line 2", "Comment", "<!-- ... -->"),
    ("Line 3", "Root Element", "<catalog>"),
    ("Line 4", "Child Element with Attributes", "<book id=\"bk101\" category=\"programming\">"),
    ("Line 5", "Element with Text Content", "<title>XML Developer's Guide</title>"),
    ("Line 7", "Element with Attribute", "<price currency=\"USD\">44.95</price>"),
    ("Line 11", "Closing Root Element", "</catalog>")
]

for line, component, example in components:
    print(f"{line:8s} → {component:30s} {example}")
````

## 2. XML Declaration

The XML declaration is **optional** but recommended. It tells the parser:
- XML version being used
- Character encoding
- Whether the document is standalone

### Syntax:
```xml
<?xml version="1.0" encoding="UTF-8" standalone="yes"?>
```

````python
# Different XML declarations

declarations = {
    "Minimal": '<?xml version="1.0"?>',
    "With Encoding (Most Common)": '<?xml version="1.0" encoding="UTF-8"?>',
    "With Standalone": '<?xml version="1.0" encoding="UTF-8" standalone="yes"?>',
    "ISO Encoding": '<?xml version="1.0" encoding="ISO-8859-1"?>',
    "XML 1.1 (Rare)": '<?xml version="1.1" encoding="UTF-8"?>'
}

print("=" * 70)
print("📋 XML DECLARATION EXAMPLES")
print("=" * 70 + "\n")

for desc, decl in declarations.items():
    print(f"{desc:30s} {decl}")

print("\n" + "=" * 70)
print("🎯 ATTRIBUTES EXPLAINED:")
print("   • version:    XML specification version (1.0 is standard)")
print("   • encoding:   Character set (UTF-8 recommended)")
print("   • standalone: yes = no external files, no = uses DTD/schema")
print("=" * 70)

print("\n💡 BEST PRACTICE: Always use UTF-8 encoding!")
````

## 3. Elements: The Building Blocks

Elements are the fundamental building blocks of XML. They consist of:
- **Opening tag**: `<element>`
- **Content**: Text, other elements, or both
- **Closing tag**: `</element>`

### Types of Elements:

````python
# Different types of XML elements

element_examples = '''
<!-- 1. Element with text content -->
<name>John Doe</name>

<!-- 2. Element with child elements -->
<person>
    <firstName>John</firstName>
    <lastName>Doe</lastName>
</person>

<!-- 3. Empty element (two ways) -->
<lineBreak></lineBreak>
<lineBreak/>

<!-- 4. Element with mixed content (text + elements) -->
<paragraph>
    This is <emphasis>really</emphasis> important!
</paragraph>

<!-- 5. Element with attributes -->
<image src="photo.jpg" width="800" height="600"/>

<!-- 6. Nested elements (hierarchy) -->
<company>
    <department name="IT">
        <employee id="101">
            <name>Alice Smith</name>
            <role>Developer</role>
        </employee>
    </department>
</company>
'''

print("=" * 70)
print("🧱 TYPES OF XML ELEMENTS")
print("=" * 70)
print(element_examples)

print("\n" + "=" * 70)
print("📝 ELEMENT RULES:")
print("   1. Must have opening and closing tags (or self-close)")
print("   2. Tag names are case-sensitive (<Name> ≠ <name>)")
print("   3. Must be properly nested (no overlapping)")
print("   4. Can contain: text, elements, or both (mixed content)")
print("   5. Empty elements can use short form: <tag/>")
print("=" * 70)
````

### Element Naming Rules

````python
print("=" * 70)
print("📛 XML ELEMENT NAMING RULES")
print("=" * 70 + "\n")

rules = [
    ("✅", "Can start with", "Letter or underscore: <name>, <_private>"),
    ("✅", "Can contain", "Letters, digits, hyphens, periods, underscores"),
    ("✅", "Valid examples", "<user-name>, <item_123>, <price.usd>"),
    ("❌", "Cannot start with", "xml, XML, Xml (reserved)"),
    ("❌", "Cannot start with", "Number or punctuation: <1st>, <-name>"),
    ("❌", "Cannot contain", "Spaces: <first name> (use <first_name> or <firstName>)"),
    ("❌", "Invalid examples", "<1user>, <first name>, <price$>"),
]

for symbol, rule_type, example in rules:
    print(f"{symbol} {rule_type:20s} → {example}")

print("\n" + "=" * 70)
print("💡 NAMING CONVENTIONS (Not required, but recommended):")
print("   • camelCase: <firstName>, <emailAddress>")
print("   • PascalCase: <FirstName>, <EmailAddress>")
print("   • kebab-case: <first-name>, <email-address>")
print("   • snake_case: <first_name>, <email_address>")
print("\n   Choose ONE style and be consistent!")
print("=" * 70)
````

## 4. Attributes

Attributes provide additional information about elements. They are name-value pairs within the opening tag.

### Syntax:
```xml
<element attribute="value">
```

````python
# Attribute examples

attribute_examples = '''
<!-- Single attribute -->
<person age="30">John</person>

<!-- Multiple attributes -->
<image src="photo.jpg" width="800" height="600" alt="Profile picture"/>

<!-- Attributes with different quote styles (both valid) -->
<book title="XML Guide" author='John Doe'/>  <!-- Mixed quotes OK -->

<!-- Attributes for metadata -->
<product id="P123" category="electronics" inStock="true">
    <name>Laptop</name>
    <price>999.99</price>
</product>

<!-- Common use: IDs and references -->
<order id="ORD001" customerId="C456" status="pending"/>
'''

print("=" * 70)
print("🏷️  XML ATTRIBUTES")
print("=" * 70)
print(attribute_examples)

print("\n" + "=" * 70)
print("📝 ATTRIBUTE RULES:")
print("   1. Must be quoted (single or double quotes)")
print("   2. No duplicate attributes in same element")
print("   3. Names follow same rules as element names")
print("   4. Values are always strings (even numbers)")
print("   5. Order doesn't matter")
print("=" * 70)
````

### When to Use Attributes vs Elements?

This is a common design question. Let's explore both approaches:

````python
print("=" * 70)
print("⚖️  ATTRIBUTES vs ELEMENTS: THE SAME DATA, TWO WAYS")
print("=" * 70)

print("\n📋 APPROACH 1: Using Attributes")
attribute_approach = '''<person id="1" firstName="John" lastName="Doe" age="30" email="john@example.com"/>'''
print(attribute_approach)

print("\n📋 APPROACH 2: Using Elements")
element_approach = '''<person>
    <id>1</id>
    <firstName>John</firstName>
    <lastName>Doe</lastName>
    <age>30</age>
    <email>john@example.com</email>
</person>'''
print(element_approach)

print("\n📋 APPROACH 3: Mixed (Common Practice)")
mixed_approach = '''<person id="1">  <!-- ID as attribute (metadata) -->
    <firstName>John</firstName>       <!-- Data as elements -->
    <lastName>Doe</lastName>
    <age>30</age>
    <email>john@example.com</email>
</person>'''
print(mixed_approach)

print("\n" + "=" * 70)
print("🎯 GUIDELINES:")
print("\n✅ USE ATTRIBUTES FOR:")
print("   • Metadata (IDs, types, categories)")
print("   • Simple values that won't expand")
print("   • Values that don't need child elements")
print("   • Data that describes the element")

print("\n✅ USE ELEMENTS FOR:")
print("   • Main content/data")
print("   • Values that might need structure later")
print("   • Multiple values of same type")
print("   • Complex data that needs attributes")

print("\n💡 GENERAL RULE:")
print("   If in doubt, use elements. They're more flexible!")
print("=" * 70)
````

## 5. Special Characters and Entity References

Some characters have special meaning in XML and must be escaped:

````python
print("=" * 70)
print("🔤 XML SPECIAL CHARACTERS & ENTITY REFERENCES")
print("=" * 70 + "\n")

print("PROBLEM: These characters are reserved in XML:\n")

special_chars = [
    ("<", "&lt;", "Less than (starts a tag)"),
    (">", "&gt;", "Greater than (ends a tag)"),
    ("&", "&amp;", "Ampersand (starts an entity)"),
    ('"', "&quot;", "Double quote (used in attributes)"),
    ("'", "&apos;", "Apostrophe (used in attributes)")
]

print(f"{'Character':^15} {'Entity':^15} {'Purpose':^35}")
print("-" * 70)
for char, entity, purpose in special_chars:
    print(f"{char:^15} {entity:^15} {purpose:^35}")

print("\n" + "=" * 70)

# Examples
print("\n📝 EXAMPLES:\n")

examples = [
    ("Wrong", "<message>5 < 10</message>", "ERROR: Looks like a tag!"),
    ("Correct", "<message>5 &lt; 10</message>", "✓ Using entity reference"),
    ("", "", ""),
    ("Wrong", '<text>She said "Hello"</text>', "ERROR: Quote closes attribute"),
    ("Correct", '<text>She said &quot;Hello&quot;</text>', "✓ Using entity reference"),
    ("", "", ""),
    ("Wrong", "<company>Smith & Sons</company>", "ERROR: & must be escaped"),
    ("Correct", "<company>Smith &amp; Sons</company>", "✓ Using entity reference")
]

for status, code, note in examples:
    if status:
        print(f"{status:8s} {code:45s} {note}")
````

````python
# Practical example: Escaping special characters

def escape_xml_chars(text):
    """Escape special XML characters in text"""
    replacements = [
        ('&', '&amp;'),   # Must be first!
        ('<', '&lt;'),
        ('>', '&gt;'),
        ('"', '&quot;'),
        ("'", '&apos;')
    ]
    for old, new in replacements:
        text = text.replace(old, new)
    return text

# Test the function
test_strings = [
    "5 < 10 and 10 > 5",
    "Smith & Jones",
    'She said "Hello" and he said \'Hi\'',
    "<html><body>Content</body></html>"
]

print("=" * 70)
print("🔧 AUTOMATIC XML CHARACTER ESCAPING")
print("=" * 70 + "\n")

for test in test_strings:
    escaped = escape_xml_chars(test)
    print(f"Original:  {test}")
    print(f"Escaped:   {escaped}")
    print(f"In XML:    <text>{escaped}</text>")
    print("-" * 70)
````

## 6. CDATA Sections

When you have a lot of special characters (like code snippets), use CDATA instead of escaping every character:

### Syntax:
```xml
<![CDATA[ ... your content here ... ]]>
```

````python
cdata_example = '''<?xml version="1.0"?>
<code-examples>
    
    <!-- Without CDATA - tedious! -->
    <example language="xml">
        &lt;html&gt;\n
            &lt;body&gt;If x &lt; 10 &amp;&amp; y &gt; 5&lt;/body&gt;\n
        &lt;/html&gt;
    </example>
    
    <!-- With CDATA - much easier! -->
    <example language="html">
        <![CDATA[
        <html>
            <body>
                if (x < 10 && y > 5) {
                    alert("Condition met!");
                }
            </body>
        </html>
        ]]>
    </example>
    
    <!-- JavaScript in CDATA -->
    <script language="javascript">
        <![CDATA[
        function compare(a, b) {
            return a < b ? -1 : (a > b ? 1 : 0);
        }
        ]]>
    </script>
    
</code-examples>'''

print("=" * 70)
print("📦 CDATA SECTIONS - For Unparsed Character Data")
print("=" * 70)
print(cdata_example)

print("\n" + "=" * 70)
print("✅ WHEN TO USE CDATA:")
print("   • Code snippets (HTML, JavaScript, SQL)")
print("   • Text with many special characters (<, >, &)")
print("   • Mathematical formulas")
print("   • Any content that looks like XML but isn't")

print("\n❌ CDATA LIMITATIONS:")
print("   • Cannot contain ']]>' (ends the CDATA section)")
print("   • Cannot be nested")
print("   • Parser ignores ALL markup inside")
print("=" * 70)
````

## 7. Comments

XML comments are just like HTML comments:

````python
comment_examples = '''<?xml version="1.0"?>
<!-- Single line comment -->

<config>
    <!--
        Multi-line comment
        Can span multiple lines
        Great for documentation
    -->
    
    <setting name="timeout">30</setting>  <!-- Inline comment -->
    
    <!-- Temporarily disable a section
    <setting name="debug">true</setting>
    <setting name="verbose">true</setting>
    -->
    
    <!-- TODO: Add error handling -->
    <errorHandler>default</errorHandler>
</config>'''

print("=" * 70)
print("💬 XML COMMENTS")
print("=" * 70)
print(comment_examples)

print("\n" + "=" * 70)
print("📝 COMMENT RULES:")
print("   ✅ Can appear almost anywhere")
print("   ✅ Can be multi-line")
print("   ✅ Syntax: <!-- comment -->")
print("   ❌ Cannot contain '--' (double hyphen)")
print("   ❌ Cannot be nested")
print("   ❌ Cannot appear before XML declaration")

print("\n💡 BEST PRACTICES:")
print("   • Document complex structures")
print("   • Explain business rules")
print("   • Mark TODO items")
print("   • Don't over-comment obvious things")
print("=" * 70)
````

## 8. Processing Instructions

Processing instructions (PIs) pass information to applications that process the XML:

````python
pi_examples = '''<?xml version="1.0"?>
<?xml-stylesheet type="text/xsl" href="style.xsl"?>
<?xml-stylesheet type="text/css" href="style.css"?>

<document>
    <?pagebreak?>  <!-- Instruction to insert page break -->
    
    <chapter>
        <?editor note="Review this section"?>
        <title>Introduction</title>
    </chapter>
    
    <?php
        // PHP code can be embedded
        echo "Hello World";
    ?>
</document>'''

print("=" * 70)
print("⚙️  PROCESSING INSTRUCTIONS (PIs)")
print("=" * 70)
print(pi_examples)

print("\n" + "=" * 70)
print("📝 PI RULES:")
print("   • Syntax: <?target instruction?>")
print("   • Target identifies the application")
print("   • Cannot use 'xml' as target (reserved)")
print("   • Common use: linking stylesheets")

print("\n🎯 COMMON USES:")
print("   • Linking XSLT stylesheets")
print("   • Linking CSS stylesheets")
print("   • Page breaks in documents")
print("   • Editor/application-specific instructions")
print("=" * 70)
````

## 9. Practical Example: Building a Complete XML Document

````python
# Let's create a complete, real-world XML document programmatically

from datetime import datetime

def create_product_xml():
    """Create a product catalog XML document"""
    
    # Create root element
    root = ET.Element('productCatalog')
    root.set('version', '1.0')
    root.set('generated', datetime.now().strftime('%Y-%m-%d %H:%M:%S'))
    
    # Add products
    products_data = [
        {
            'id': 'P001',
            'category': 'Electronics',
            'name': 'Laptop',
            'price': 999.99,
            'currency': 'USD',
            'inStock': True,
            'description': 'High-performance laptop with 16GB RAM'
        },
        {
            'id': 'P002',
            'category': 'Electronics',
            'name': 'Smartphone',
            'price': 699.99,
            'currency': 'USD',
            'inStock': True,
            'description': 'Latest model with 5G support & 128GB storage'
        },
        {
            'id': 'P003',
            'category': 'Books',
            'name': 'XML Mastery',
            'price': 49.99,
            'currency': 'USD',
            'inStock': False,
            'description': 'Complete guide to XML & web services'
        }
    ]
    
    for prod_data in products_data:
        product = ET.SubElement(root, 'product')
        product.set('id', prod_data['id'])
        product.set('category', prod_data['category'])
        
        name = ET.SubElement(product, 'name')
        name.text = prod_data['name']
        
        price = ET.SubElement(product, 'price')
        price.set('currency', prod_data['currency'])
        price.text = str(prod_data['price'])
        
        stock = ET.SubElement(product, 'inStock')
        stock.text = str(prod_data['inStock']).lower()
        
        desc = ET.SubElement(product, 'description')
        desc.text = prod_data['description']
    
    return root

# Generate the XML
catalog = create_product_xml()

# Pretty print
xml_string = ET.tostring(catalog, encoding='unicode')
dom = minidom.parseString(xml_string)
pretty_xml = dom.toprettyxml(indent='    ')

print("=" * 70)
print("🏭 GENERATED XML DOCUMENT")
print("=" * 70)
print(pretty_xml)

# Save to file
with open('product_catalog.xml', 'w', encoding='utf-8') as f:
    f.write(pretty_xml)

print("\n✅ XML document saved to: product_catalog.xml")
````
