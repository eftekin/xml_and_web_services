## Part 1: JSON Fundamentals

### What is JSON?

**JSON** (JavaScript Object Notation) is a lightweight data interchange format that's:
- Easy for humans to read and write
- Easy for machines to parse and generate
- Language-independent (despite the name)
- The de facto standard for Web APIs

### JSON vs XML: Quick Comparison

```python
import json
import xml.etree.ElementTree as ET
from xml.dom import minidom
import requests
from tabulate import tabulate

print("=" * 80)
print("📊 JSON vs XML: THE SAME DATA")
print("=" * 80)

# JSON representation
json_data = {
    "user": {
        "id": 1,
        "name": "John Doe",
        "email": "john@example.com",
        "active": True,
        "roles": ["admin", "editor"],
        "address": {
            "street": "123 Main St",
            "city": "New York",
            "zip": "10001"
        }
    }
}

print("\n📗 JSON FORMAT:")
print(json.dumps(json_data, indent=2))

# XML representation
xml_data = '''<?xml version="1.0" encoding="UTF-8"?>
<user>
    <id>1</id>
    <name>John Doe</name>
    <email>john@example.com</email>
    <active>true</active>
    <roles>
        <role>admin</role>
        <role>editor</role>
    </roles>
    <address>
        <street>123 Main St</street>
        <city>New York</city>
        <zip>10001</zip>
    </address>
</user>'''

print("\n📙 XML FORMAT:")
print(xml_data)

# Size comparison
json_size = len(json.dumps(json_data))
xml_size = len(xml_data)

print("\n" + "=" * 80)
print("📏 SIZE COMPARISON:")
print(f"   JSON: {json_size} bytes")
print(f"   XML:  {xml_size} bytes")
print(f"   JSON is {((xml_size - json_size) / xml_size * 100):.1f}% smaller!")
print("=" * 80)
```

### JSON Syntax Rules

JSON has only **6 data types**:

```python
# JSON Data Types

json_types_example = {
    "string": "Hello World",           # Text in double quotes
    "number_int": 42,                  # Integer
    "number_float": 3.14,              # Floating point
    "boolean": True,                    # true or false (lowercase in JSON)
    "null_value": None,                 # null (None in Python)
    "array": [1, 2, 3, "mixed", True], # Ordered list
    "object": {                         # Key-value pairs
        "nested": "value",
        "count": 10
    }
}

print("=" * 80)
print("🔤 JSON DATA TYPES")
print("=" * 80 + "\n")

print("1️⃣  STRING: Text in double quotes")
print(f'   {json.dumps({"example": "Hello World"})}')

print("\n2️⃣  NUMBER: Integer or floating-point")
print(f'   {json.dumps({"int": 42, "float": 3.14})}')

print("\n3️⃣  BOOLEAN: true or false (lowercase!)")
print(f'   {json.dumps({"active": True, "deleted": False})}')

print("\n4️⃣  NULL: Represents absence of value")
print(f'   {json.dumps({"optional": None})}')

print("\n5️⃣  ARRAY: Ordered list of values")
print(f'   {json.dumps({"numbers": [1, 2, 3], "mixed": ["a", 1, True]})}')

print("\n6️⃣  OBJECT: Key-value pairs (like Python dict)")
print(f'   {json.dumps({"user": {"name": "John", "age": 30}}, indent=2)}')

print("\n" + "=" * 80)
print("📝 JSON SYNTAX RULES:")
print("   ✅ Keys must be strings in double quotes")
print("   ✅ String values in double quotes (not single)")
print("   ✅ No trailing commas")
print("   ✅ No comments allowed")
print("   ✅ Boolean: true/false (lowercase)")
print("   ✅ Null: null (lowercase)")
print("=" * 80)
```

### Working with JSON in Python

```python
# JSON operations in Python

print("=" * 80)
print("🐍 JSON IN PYTHON")
print("=" * 80)

# 1. Python object to JSON string (serialization)
print("\n1️⃣  PYTHON DICT → JSON STRING (dumps)\n")
python_dict = {
    "name": "Alice",
    "age": 25,
    "skills": ["Python", "JavaScript"]
}
json_string = json.dumps(python_dict, indent=2)
print(f"Python dict: {python_dict}")
print(f"\nJSON string:\n{json_string}")

# 2. JSON string to Python object (deserialization)
print("\n" + "-" * 80)
print("\n2️⃣  JSON STRING → PYTHON DICT (loads)\n")
json_text = '{"product": "Laptop", "price": 999.99, "inStock": true}'
parsed_data = json.loads(json_text)
print(f"JSON string: {json_text}")
print(f"\nParsed dict: {parsed_data}")
print(f"Access price: ${parsed_data['price']}")

# 3. Pretty printing
print("\n" + "-" * 80)
print("\n3️⃣  PRETTY PRINTING\n")
complex_data = {
    "users": [
        {"id": 1, "name": "Alice", "role": "admin"},
        {"id": 2, "name": "Bob", "role": "user"}
    ],
    "total": 2
}
print("Compact:")
print(json.dumps(complex_data))
print("\nPretty (indent=2):")
print(json.dumps(complex_data, indent=2))

# 4. Type conversions
print("\n" + "-" * 80)
print("\n4️⃣  PYTHON ↔ JSON TYPE MAPPING\n")
type_mapping = [
    ("dict", "object", "{ }"),
    ("list, tuple", "array", "[ ]"),
    ("str", "string", '"text"'),
    ("int, float", "number", "123"),
    ("True, False", "true, false", "boolean"),
    ("None", "null", "null")
]

print(f"{'Python Type':20s} {'JSON Type':15s} {'Example':15s}")
print("-" * 50)
for py_type, json_type, example in type_mapping:
    print(f"{py_type:20s} {json_type:15s} {example:15s}")

print("\n" + "=" * 80)
```

## Part 2: REST Architecture

### What is REST?

**REST** (Representational State Transfer) is an architectural style for designing networked applications. It's not a protocol or standard, but a set of constraints.

### REST Principles (Constraints)

```python
print("=" * 80)
print("🏛️  REST ARCHITECTURAL CONSTRAINTS (by Roy Fielding, 2000)")
print("=" * 80 + "\n")

rest_principles = [
    (
        "1. Client-Server",
        "Separation of concerns: UI separate from data storage",
        "Benefit: Independent evolution of client and server"
    ),
    (
        "2. Stateless",
        "Each request contains all information needed",
        "Benefit: Scalability, no session state on server"
    ),
    (
        "3. Cacheable",
        "Responses must define themselves as cacheable or not",
        "Benefit: Improved performance, reduced server load"
    ),
    (
        "4. Uniform Interface",
        "Standardized way to communicate (HTTP methods, URIs)",
        "Benefit: Simplicity, discoverability"
    ),
    (
        "5. Layered System",
        "Client can't tell if connected to end server or intermediary",
        "Benefit: Load balancing, security, caching layers"
    ),
    (
        "6. Code on Demand*",
        "Server can send executable code (JavaScript)",
        "Benefit: Extend client functionality (*optional)"
    )
]

for principle, description, benefit in rest_principles:
    print(f"\n{principle}")
    print(f"  📝 {description}")
    print(f"  💡 {benefit}")

print("\n" + "=" * 80)
```

### HTTP Methods (Verbs)

REST uses HTTP methods to indicate the desired action:

```python
print("=" * 80)
print("🌐 HTTP METHODS (RESTful APIs)")
print("=" * 80 + "\n")

http_methods = [
    ("GET", "Read/Retrieve", "Get a resource", "Idempotent, Safe", "GET /users/1"),
    ("POST", "Create", "Create new resource", "Not idempotent", "POST /users"),
    ("PUT", "Update/Replace", "Replace entire resource", "Idempotent", "PUT /users/1"),
    ("PATCH", "Update/Modify", "Partial update", "Not idempotent", "PATCH /users/1"),
    ("DELETE", "Delete", "Remove resource", "Idempotent", "DELETE /users/1"),
    ("HEAD", "Metadata", "Get headers only", "Idempotent, Safe", "HEAD /users/1"),
    ("OPTIONS", "Options", "Get allowed methods", "Idempotent, Safe", "OPTIONS /users")
]

print(f"{'Method':10s} {'Purpose':15s} {'Description':20s} {'Properties':20s} {'Example':15s}")
print("-" * 80)
for method, purpose, desc, props, example in http_methods:
    print(f"{method:10s} {purpose:15s} {desc:20s} {props:20s} {example:15s}")

print("\n" + "=" * 80)
print("📝 KEY TERMS:")
print("   • Idempotent: Multiple identical requests have same effect as one")
print("   • Safe: Doesn't modify resources (read-only)")
print("=" * 80)
```

### HTTP Status Codes

```python
print("=" * 80)
print("📊 HTTP STATUS CODES (Common Ones)")
print("=" * 80 + "\n")

status_codes = {
    "2xx - Success": [
        ("200", "OK", "Request succeeded"),
        ("201", "Created", "Resource created successfully"),
        ("204", "No Content", "Success but no content to return")
    ],
    "3xx - Redirection": [
        ("301", "Moved Permanently", "Resource has new permanent URI"),
        ("304", "Not Modified", "Cached version is still valid")
    ],
    "4xx - Client Errors": [
        ("400", "Bad Request", "Malformed request"),
        ("401", "Unauthorized", "Authentication required"),
        ("403", "Forbidden", "Authenticated but not authorized"),
        ("404", "Not Found", "Resource doesn't exist"),
        ("429", "Too Many Requests", "Rate limit exceeded")
    ],
    "5xx - Server Errors": [
        ("500", "Internal Server Error", "Generic server error"),
        ("502", "Bad Gateway", "Invalid response from upstream"),
        ("503", "Service Unavailable", "Server temporarily unavailable")
    ]
}

for category, codes in status_codes.items():
    print(f"\n{category}")
    print("-" * 60)
    for code, name, description in codes:
        print(f"  {code} {name:25s} → {description}")

print("\n" + "=" * 80)
print("💡 QUICK GUIDE:")
print("   2xx → Success")
print("   3xx → Redirect somewhere else")
print("   4xx → You (client) made a mistake")
print("   5xx → Server made a mistake")
print("=" * 80)
```

## Part 3: Practical REST API Usage

Let's consume a real REST API using Python!

```python
# Using JSONPlaceholder - a fake REST API for testing

BASE_URL = "https://jsonplaceholder.typicode.com"

print("=" * 80)
print("🌐 CONSUMING REST APIs with Python")
print("=" * 80)
print(f"\nAPI: JSONPlaceholder ({BASE_URL})")
print("A free REST API for testing and prototyping\n")

# 1. GET - Retrieve data
print("\n" + "="*80)
print("1️⃣  GET REQUEST - Retrieve a user")
print("="*80)

response = requests.get(f"{BASE_URL}/users/1")
print(f"\nURL: {response.url}")
print(f"Status Code: {response.status_code} {response.reason}")
print(f"\nResponse:")
user = response.json()
print(json.dumps(user, indent=2))

# 2. GET with query parameters
print("\n" + "="*80)
print("2️⃣  GET with Parameters - Filter posts by user")
print("="*80)

params = {"userId": 1}
response = requests.get(f"{BASE_URL}/posts", params=params)
posts = response.json()

print(f"\nURL: {response.url}")
print(f"Found {len(posts)} posts by user 1")
print(f"\nFirst post:")
print(json.dumps(posts[0], indent=2))

# 3. POST - Create new resource
print("\n" + "="*80)
print("3️⃣  POST REQUEST - Create new post")
print("="*80)

new_post = {
    "title": "Learning REST APIs",
    "body": "This is my first post about REST!",
    "userId": 1
}

response = requests.post(f"{BASE_URL}/posts", json=new_post)
created_post = response.json()

print(f"\nStatus Code: {response.status_code} {response.reason}")
print(f"\nCreated post:")
print(json.dumps(created_post, indent=2))

# 4. PUT - Update (replace) resource
print("\n" + "="*80)
print("4️⃣  PUT REQUEST - Update entire post")
print("="*80)

updated_post = {
    "id": 1,
    "title": "Updated Title",
    "body": "This replaces the entire post",
    "userId": 1
}

response = requests.put(f"{BASE_URL}/posts/1", json=updated_post)
result = response.json()

print(f"\nStatus Code: {response.status_code}")
print(f"\nUpdated post:")
print(json.dumps(result, indent=2))

# 5. PATCH - Partial update
print("\n" + "="*80)
print("5️⃣  PATCH REQUEST - Update only title")
print("="*80)

partial_update = {
    "title": "Only title changed"
}

response = requests.patch(f"{BASE_URL}/posts/1", json=partial_update)
result = response.json()

print(f"\nStatus Code: {response.status_code}")
print(f"\nPartially updated post:")
print(json.dumps(result, indent=2))

# 6. DELETE - Remove resource
print("\n" + "="*80)
print("6️⃣  DELETE REQUEST - Remove a post")
print("="*80)

response = requests.delete(f"{BASE_URL}/posts/1")

print(f"\nStatus Code: {response.status_code}")
print(f"Response: {response.text if response.text else '(empty - success)'}")

print("\n" + "="*80)
print("✅ All REST operations demonstrated!")
print("="*80)
```

### Error Handling and Best Practices

```python
# Professional API consumption with error handling

def fetch_user(user_id):
    """Fetch user with comprehensive error handling"""
    try:
        response = requests.get(
            f"{BASE_URL}/users/{user_id}",
            timeout=5  # Always set timeout!
        )
        
        # Raise exception for bad status codes
        response.raise_for_status()
        
        return response.json()
        
    except requests.exceptions.Timeout:
        print("❌ Error: Request timed out")
    except requests.exceptions.ConnectionError:
        print("❌ Error: Could not connect to server")
    except requests.exceptions.HTTPError as e:
        print(f"❌ HTTP Error: {e}")
    except requests.exceptions.JSONDecodeError:
        print("❌ Error: Invalid JSON response")
    except Exception as e:
        print(f"❌ Unexpected error: {e}")
    
    return None

# Test the function
print("=" * 80)
print("🛡️  PROFESSIONAL API CONSUMPTION")
print("=" * 80)

print("\n1. Valid request:")
user = fetch_user(1)
if user:
    print(f"✅ Successfully fetched: {user['name']}")

print("\n2. Invalid request (non-existent user):")
user = fetch_user(99999)

print("\n" + "=" * 80)
print("💡 BEST PRACTICES:")
print("   ✓ Always set timeouts")
print("   ✓ Handle all exception types")
print("   ✓ Check status codes")
print("   ✓ Validate response data")
print("   ✓ Use HTTPS in production")
print("   ✓ Implement retry logic for transient failures")
print("   ✓ Log requests and responses")
print("   ✓ Respect rate limits")
print("=" * 80)
```

## Part 4: RESTful URI Design

### Best Practices for API Endpoints

```python
print("=" * 80)
print("🎯 RESTful URI DESIGN PATTERNS")
print("=" * 80 + "\n")

print("✅ GOOD PRACTICES:\n")

good_examples = [
    ("Collection", "GET /users", "Get all users"),
    ("Resource", "GET /users/123", "Get specific user"),
    ("Sub-collection", "GET /users/123/posts", "Get user's posts"),
    ("Sub-resource", "GET /users/123/posts/456", "Get specific post of user"),
    ("Filtering", "GET /users?role=admin&active=true", "Filter with query params"),
    ("Sorting", "GET /users?sort=name&order=desc", "Sort results"),
    ("Pagination", "GET /users?page=2&limit=20", "Paginate results"),
    ("Partial response", "GET /users?fields=id,name,email", "Select specific fields")
]

for pattern, endpoint, description in good_examples:
    print(f"{pattern:18s}  {endpoint:45s}  {description}")

print("\n" + "="*80)
print("❌ BAD PRACTICES (Avoid These):\n")

bad_examples = [
    ("Verbs in URI", "GET /getUsers", "Use nouns, not verbs"),
    ("CRUD in URI", "POST /createUser", "HTTP method indicates action"),
    ("Mixed case", "GET /Users/getProfile", "Use lowercase"),
    ("File extensions", "GET /users.json", "Use Accept header instead"),
    ("Deep nesting", "GET /a/b/c/d/e/f", "Max2-3 levels"),
    ("Underscores", "GET /user_profiles", "Use hyphens: /user-profiles")
]

for anti_pattern, bad_endpoint, why in bad_examples:
    print(f"{anti_pattern:18s}  {bad_endpoint:45s}  ❌ {why}")

print("\n" + "="*80)
print("📋 URI DESIGN RULES:")
rules = [
    "Use nouns, not verbs (GET /users, not /getUsers)",
    "Use plural names for collections (/users, not /user)",
    "Use hyphens, not underscores (/user-profiles)",
    "Use lowercase letters only",
    "Don't use file extensions",
    "Keep URIs short and readable",
    "Use query parameters for filtering/sorting",
    "Version your API (/v1/users, /v2/users)"
]
for i, rule in enumerate(rules, 1):
    print(f"   {i}. {rule}")
print("="*80)
```
